```python
data = pd.read_csv("pca_exercise2.csv")

print("Dataset:")
print(data)

print("\nDataset Shape:")
print(data.shape)
```

```python
print("Variables:")
print(data['Variable'])

print("\nPC1 Loadings:")
print(data['PC1'])

print("\nPC2 Loadings:")
print(data['PC2'])

print("\nPC3 Loadings:")
print(data['PC3'])
```

```python
index = data['PC1'].abs().idxmax()

variable = data.loc[index, 'Variable']
loading = data.loc[index, 'PC1']

print("PC1 Analysis")
print("Variable:", variable)
print("Loading:", loading)
```

```python
index = data['PC2'].abs().idxmax()

variable = data.loc[index, 'Variable']
loading = data.loc[index, 'PC2']

print("PC2 Analysis")
print("Variable:", variable)
print("Loading:", loading)
```

```python
index = data['PC3'].abs().idxmax()

variable = data.loc[index, 'Variable']
loading = data.loc[index, 'PC3']

print("PC3 Analysis")
print("Variable:", variable)
print("Loading:", loading)
```

```python
for pc in ['PC1', 'PC2', 'PC3']:

    index = data[pc].abs().idxmax()

    variable = data.loc[index, 'Variable']
    loading = data.loc[index, pc]

    print(pc, "->", variable, "| Loading:", loading)
```

```python
import numpy as np

x = data[['PC1', 'PC2', 'PC3']].values

print("Shape of x:", x.shape)
print(x)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(
    x[:, 0],
    x[:, 1],
    c=x[:, 0],
    cmap='plasma',
    s=80
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA 2D - Component Loadings")

plt.colorbar(label="PC1 Loading")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    x[:, 0],
    x[:, 1],
    x[:, 2],
    c=x[:, 0],
    cmap='plasma',
    s=80
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

ax.set_title("PCA 3D - Component Loadings")

plt.colorbar(
    scatter,
    label="PC1 Loading"
)

plt.show()
```
